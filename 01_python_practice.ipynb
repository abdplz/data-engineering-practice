## Q.1
Two lists contain product codes. Find the codes that appear in only one list.

```python
delivery_a = [12, 18, 21, 18, 25, 30]
delivery_b = [18, 22, 25, 32, 35]

# to find unique values
unique_codes = sorted(set(delivery_a) ^ set(delivery_b))
unique_codes
```

Sets remove duplicates. The symmetric difference keeps the values that are not shared.

## Q.2
Create a list of squares of odd numbers from 1 to 15. First use a loop, then a list comprehension.

```python
squares = []
for number in range(1, 16):
    if number % 2 != 0:
        squares.append(number ** 2)
print(squares)
```

Alternative way:

```python
squares = [number ** 2 for number in range(1, 16) if number % 2 != 0]
squares
```

## Q.4
Write a function that returns only positive numbers from a list.

values = [-8, 3, 0, 12, -2, 7]

```python
#function
def positive_numbers(num):
    return [num for num in num if num > 0]
```

```python
values = [-8, 3, 0, 12, -2, 7]
positive_numbers(values)
```

## Q.9
Write a function using map that returns the length of each word in a sentence. (For instance: Small steps make Python easier)

```python
#function for String
def word_lengths(sentence):
    words = sentence.split()
    return list(map(len, words))
```

```python
word_lengths("Small steps make Python easier")
```

## Q.13

Calculate each product's stock value using dictionary comprehension.

stock = {"cups": 25, "plates": None, "forks": 0, "glasses": 18}

prices = {"cups": 2.50, "plates": 6.00, "forks": 1.50, "glasses": 4.00}

```python
stock = {"cups": 25, "plates": None, "forks": 0, "glasses": 18}
prices = {"cups": 2.50, "plates": 6.00, "forks": 1.50, "glasses": 4.00}

stock_value = {}

for product, quantity in stock.items():
    if quantity is not None:
        if quantity > 0:
            stock_value[product] = quantity * prices[product]

stock_value
```

## Q.15
Find the highest and lowest daily sales for each branch. Then identify the branches containing the overall highest and lowest values.

```python
sales = {
    "centre": [230, 310, 280, 420],
    "airport": [510, 460, 620, 550],
    "harbour": [190, 260, 340, 290]
}
for branch, amounts in sales.items():
    print(branch, "min:", min(amounts), "max:", max(amounts))

highest_branch = max(sales, key=lambda branch: max(sales[branch]))
lowest_branch = min(sales, key=lambda branch: min(sales[branch]))
```

```python
print("Highest:", highest_branch, max(sales[highest_branch]))
print("Lowest:", lowest_branch, min(sales[lowest_branch]))
```

## Q.16
Find the most frequent letter in the sentence. Ignore spaces and letter case.

```python
#To remove , from string
sentence = "Small steps make progress"
letters = sentence.lower().replace(" ", "")
letters
```

```python
counts = {}
for letter in letters:
    if letter in counts:
        counts[letter] += 1
    else:
        counts[letter] = 1
        
print(counts)
```

```python
#max value in dict
most_common = max(counts, key=counts.get)

[most_common, counts[most_common]]
```
